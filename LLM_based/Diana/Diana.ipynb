```python
from tabula.tabula import Tabula
import pandas as pd
import torch
```

```python
file_path = '../../Data/data.csv'
data = pd.read_csv(file_path)
data.head()
```

```python
columns = ['region', 'customer_age_class', 'basket_package_price_disc',   
           'internet_package_price_disc', 'basket_days_until_contract_end',
           'count_previous_cancellations_0_18', 'internet_product_name',
           'flag_churn']
data = data[columns]
```

```python
data.head()
```

```python
# gpt model
categorical_columns = ['region', 'customer_age_class', 'internet_product_name']
model = Tabula(llm='gpt2', experiment_dir = "adult_training", batch_size=16, epochs=5, categorical_columns = categorical_columns)
```

```python
model.fit(data)
```

```python
synthetic_data_diana = model.sample(n_samples=10000)
```

```python
synthetic_data_diana
```

```python
synthetic_data_diana.to_csv('diana_data.csv', index=False)
```

```python
#Shows that all the synthetic values are from the original value set
set1 = set(synthetic_data_diana['internet_product_name'].unique())
set2 = set(data['internet_product_name'].unique())
list(set1-set2)
```
