## TabuLa

```python
from tabula.tabula import Tabula
import pandas as pd
import torch
```

```python
file_path = '../../Data/data.csv'
data = pd.read_csv(file_path)
```

```python
pd.set_option('display.max_columns', None)
data.head()
```

```python
churn_1_sample = data[data['flag_churn'] == 1].sample(n=10000, random_state=42)
churn_0_sample = data[data['flag_churn'] == 0].sample(n=10000, random_state=42)
balanced_sample = pd.concat([churn_1_sample, churn_0_sample])
balanced_sample = balanced_sample.sample(frac=1, random_state=42).reset_index(drop=True)
```

```python
categorical_columns = ['region', 'customer_age_class', 'basket_product_category', 'internet_product_name']
```

```python
model = Tabula(llm='distilgpt2', experiment_dir='tabula_training', batch_size=32, epochs=100, categorical_columns=categorical_columns)
```

```python
model.fit(balanced_sample)
```

```python
synthetic_data = model.sample(n_samples=50)
```

```python
synthetic_data
```

```python
synthetic_data.to_csv("tabula.csv", index=False)
```
